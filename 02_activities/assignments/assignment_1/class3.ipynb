```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy as scp
import PIL
import requests
```

```python
np.random.seed(613)

x = np.arange(50) # array of integers from 0-49
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.plot(x,y1)
ax.plot(x,y2)
plt.show()
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.plot(x,y1,label='Person 1')
ax.plot(x,y2,label='Person 2')
ax.legend(loc='best')
#ax.legend()
plt.show()
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.plot(x,y1,label='Person 1')
ax.plot(x,y2,label='Person 2')
ax.legend(loc='upper left',
          frameon = True,
          shadow = True,
          fontsize = 12,
          ncol=1,
          bbox_to_anchor = (1,1))
#ax.legend(loc='best')
plt.show()
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.scatter(x,y1,label='Person 1')
ax.scatter(x,y2,label='Person 2')
ax.legend(loc='lower right')
ax.text(x[1],y1[1],'This value is important',ha='center',va='center',color='red',size=12)
ax.text(x[10],y1[10],'This value is not',ha='center',va='center',color='#1e421e',size=8)
plt.show()
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(6,6))
ax.axis([0,10,0,10])

ax.text(1,5,". Data:(1,5) \n at x=1 and y=5", transform=ax.transData)
ax.text(0.5,0.1,". Axis:(0.5,0.1) \n at halfway along x, and tenth way \n along y in the axis object", transform=ax.transAxes)
ax.text(0.2,0.2,". Figure:(0.2,0.2) \n at 1 fifth of the way \n in x and y \n releative to \n the whole figure", transform=fig.transFigure)
plt.show()
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.scatter(x,y1,label='Person 1')
ax.scatter(x,y2,label='Person 2')
ax.legend(loc='lower right')
ax.annotate('important point',xy=(10,95),xytext=(20,94),
           arrowprops=dict(facecolor='black',width=0.001,headwidth=5))
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.scatter(x,y1,label='Person 1')
ax.scatter(x,y2,label='Person 2')
i = np.random.randint(0,50,1)
ax.legend(loc='lower right')
ax.annotate('important point',xy=(x[i]+1,y1[i]),xytext=(x[i]+5,y1[i]+5),
           arrowprops=dict(facecolor='black',arrowstyle='fancy',color='pink'),color='green')
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.scatter(x,y1,label='Person 1')
ax.scatter(x,y2,label='Person 2')
ax.legend(loc='lower right')

ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter())
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.scatter(x,y1,label='Person 1')
ax.scatter(x,y2,label='Person 2')
ax.legend(loc='lower right')

ax.set_yticks([])
ax.set_xticklabels([])
```

```python
fig,ax = plt.subplots(nrows=1,ncols=2,figsize=(10,3))
ax[0].scatter(x,y1,label='Person 1')
ax[0].scatter(x,y2,label='Person 2')
ax[0].legend(loc='lower right')
ax[0].xaxis.set_major_locator(plt.MaxNLocator(10))

ax[1].scatter(x,y1,label='Person 1')
ax[1].scatter(x,y2,label='Person 2')
ax[1].legend(loc='lower right')
ax[1].xaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.scatter(x,y1,label='Person 1')
ax.scatter(x,y2,label='Person 2')
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))

plt.xticks(rotation=45,ha='right')
```
