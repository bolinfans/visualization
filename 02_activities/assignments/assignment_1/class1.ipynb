```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy as scp
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.scatter(x,y)
ax.set_xlabel('X Label')
ax.set_ylabel('Y Label')
ax.set_title('Scatter')
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.bar(x,y)
ax.set_xlabel('X Label')
ax.set_ylabel('Y Label')
ax.set_title('Bar Plot')
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.plot(x,y)
ax.set_xlabel('X Label')
ax.set_ylabel('Y Label')
ax.set_title('Line Plot')
```

```python
fig,ax = plt.subplots(nrows=1,ncols=2,figsize=(9,3),sharex=True,sharey=True)
ax[0].hist(y,bins=20,edgecolor='black')
ax[0].set_xlabel('Total growth')
ax[0].set_ylabel('Count')
ax[1].hist(y,edgecolor='black')
ax[1].set_xlabel('Total growth')
ax[0].set_title('20 bins')
ax[1].set_title('default bins')
plt.suptitle('Histograms')
fig.tight_layout()
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.plot(x,y)
ax.set_xlabel('Year since start')
ax.set_ylabel('Total growth')
ax.set_title('Total growth over time')
fig.tight_layout()
```

```python
font1={'family':'sans-serif',
       'color':'blue',
       'size':20}

font2={'family':'monospace',
       'color':'green',
       'size':14}

fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.plot(x,y)
ax.set_xlabel('Year since start',fontdict = font2)
ax.set_ylabel('Total growth',fontdict = font2)
ax.set_title('Total growth over time',fontdict = font1)
fig.tight_layout()
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.scatter(x,y,marker='+',color='pink',s=100)
plt.show()
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.plot(x,y,linestyle='--',linewidth=2,marker='*',c='#C8E087')
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.plot(x,y,
        linestyle='--',linewidth=2,color='#DDFCAD',
        marker='*',markeredgecolor='#000000',markerfacecolor='#F5759F',markersize=12)
```

```python
fig,ax = plt.subplots(nrows=1,ncols=1,figsize=(5,3))
ax.scatter(x,y,marker='*',edgecolor='#000000',facecolor='#F5759F',s=100)
ax.grid(axis='y',c='#8D7993',linestyle=':',linewidth=1)
```
