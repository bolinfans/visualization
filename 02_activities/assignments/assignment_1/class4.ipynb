```python
from PIL import Image
import requests
from io import BytesIO
```

```python
url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "image/*,*/*;q=0.8",
    "Referer": "https://en.wikipedia.org/"
}
```

```python
response = requests.get(url,headers=headers)
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

```python
image = np.array(image)
```

```python
fig,ax = plt.subplots(nrows=1,ncols=5,figsize=(7,3),sharex=True,sharey=True)
ax[0].imshow(image)
ax[1].imshow(image[:,:,0],cmap='Reds')
ax[2].imshow(image[:,:,1],cmap='Greens')
ax[3].imshow(image[:,:,2],cmap='Blues')
ax[4].imshow(image[:,:,3],cmap='Greys')

ax[0].set_title('image')
ax[1].set_title('reds')
ax[2].set_title('greens')
ax[3].set_title('blues')
ax[4].set_title('mask')

plt.xticks([])
plt.yticks([])
```
